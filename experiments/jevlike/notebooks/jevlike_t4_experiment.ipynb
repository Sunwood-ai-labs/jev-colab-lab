# Jevlike tiny option-attention: T4 short run

This notebook downloads the task-pinned measurement wrapper from the public experiment branch. The wrapper fetches the official `vinnylarouge/jevlike` repository at commit `94f5fd1b0b11d52bbdfdf4e0ee6aa96b568f8452`, installs it with `uv`, trains the upstream `TinyScorer` on generated synthetic data, and records prediction probabilities, timings, and peak VRAM.

This is an independent Jev-like starter experiment, not the private TypeSafe Jev model.

```python
from pathlib import Path
import json
import subprocess
import sys
import urllib.request

TASK_REPOSITORY = 'https://raw.githubusercontent.com/Sunwood-ai-labs/jev-colab-lab/codex/jevlike-tiny-attention/experiments/jevlike/scripts/run_experiment.py'
runner = Path('/content/run_jevlike_experiment.py')
urllib.request.urlretrieve(TASK_REPOSITORY, runner)
result_dir = Path('/content/jevlike-results')
command = [sys.executable, str(runner), '--device', 'cuda', '--output-dir', str(result_dir)]
subprocess.run(command, check=True)
result = json.loads((result_dir / 'result.json').read_text())
result['status'], result['environment']
```

```python
summary = {
    'gpu': result['environment'].get('gpu_name'),
    'device': result['environment']['device'],
    'test': result['metrics']['test'],
    'shuffled_context_control': result['metrics']['shuffled_context_control'],
    'training_seconds': result['timings']['training_seconds'],
    'checkpoint_load_seconds': result['timings']['checkpoint_load_seconds'],
    'first_batch_inference_seconds': result['timings']['first_batch_inference_seconds'],
    'steady_state_ms': result['timings']['steady_state_ms'],
    'peak_vram': result['peak_vram'],
    'prediction': result['prediction'],
}
print(json.dumps(summary, ensure_ascii=False, indent=2))
```
